```python
class ModelReflexAgent:
    def __init__(self, desired_temperature):
        self.target_temperature = desired_temperature
        self.last_action = None
    
    def sensor(self, temperature):
        self.current_temperature = temperature
    
    def decide_action(self):
        action = None
        if self.current_temperature > self.target_temperature:
            action = "Turn on the AC"
        else:
            action = "Turn off the AC"
        
        if self.last_action == action:
            action = "No change required"
        
        self.last_action = action
        return action
    
    def actuator(self):
        action = self.decide_action()
        print(f"Temperature: {self.current_temperature}°C → Action: {action}")


room_temperatures = {
    "Living Room": 20,
    "Drawing Room": 22,
    "Bedroom": 18
}

agent = ModelReflexAgent(23)

for room, temperature in room_temperatures.items():
    print(f"{room}: ", end="")
    agent.sensor(temperature)
    agent.actuator()

print("\nChecking an individual room:")
agent.sensor(25)
agent.actuator()
```

```python
def bfs_without_queue(graph, start):
    seen = []

    def explore(levels):
        if not levels:
            return
        upcoming = []
        for val in levels:
            if val not in seen:
                print(val, end=" ")
                seen.append(val)
                upcoming.extend(graph[val])
        explore(upcoming)

    explore([start])

print("BFS without Queue & Node:")
bfs_without_queue(graph, 'A')


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G','H'],
    'E': [],
    'F': ['I','K'],
    'G': [],
    'H': ['L'],
    'I': [],
    'K': ['M'],
    'L': [],
    'M': []
}
```

```python

def bfs_with_queue(graph, start):
    seen = []
    bag = [start]

    while bag:
        item = bag.pop(0)
        if item not in seen:
            print(item, end=" ")
            seen.append(item)
            for nxt in graph[item]:
                if nxt not in seen:
                    bag.append(nxt)

print("BFS with Queue:")
bfs_with_queue(graph, 'A')


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G','H'],
    'E': [],
    'F': ['I','K'],
    'G': [],
    'H': ['L'],
    'I': [],
    'K': ['M'],
    'L': [],
    'M': []
}
```

## 4. data pre-processing 

#### 4.1 manage null values 


```python

print(df.isnull().sum())


df = df.dropna()

print(df.isnull().sum())
```

```python

cols = ['magnitude', 'depth_km', 'region', 'seismic_station_reading']


df = df.dropna(subset=cols)


print(df.isnull().sum())
```

```python

df['magnitude'] = df['magnitude'].fillna(df['magnitude'].mean())
df['depth_km'] = df['depth_km'].fillna(df['depth_km'].mean())
df['seismic_station_reading'] = df['seismic_station_reading'].fillna(df['seismic_station_reading'].mean())


df['region'] = df['region'].fillna(df['region'].mode()[0])
```

```python

def fillnaObjectMode(cols):
    for col in cols:
        df[col] = df[col].fillna(df[col].mode()[0])


string_columns = ['region']


fillnaObjectMode(string_columns)
```

```python
# ---- Function for Int Type Columns (fill with mode) ----
def fillnaIntMode(cols):
    for col in cols:
        df[col] = df[col].fillna(df[col].mode()[0])

# ---- Integer columns list for opr dataset ----
int_columns = ['warning_issued']

# ---- Call the function ----
fillnaIntMode(int_columns)
```

```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# --- Read Dataset ---
df = pd.read_csv("data.csv")

# --- Select X and y ---
X = df[["latitude", "longitude", "magnitude", "depth_km", "seismic_station_reading"]]
y = df["warning_issued"]

# --- Handle Missing Values ---
X = X.fillna(X.mean())

# --- Train-Test Split (70% train, 30% test) ---
train_X, test_X, train_y, test_y = train_test_split(
    X, 
    y, 
    test_size=0.3,
    shuffle=False
)

# --- Train SVM Model ---
model_svc = SVC()
model_svc.fit(train_X, train_y)

# --- Prediction ---
pred_y = model_svc.predict(test_X)

# --- Accuracy ---
accuracy = accuracy_score(test_y, pred_y)

print("SVM Model Accuracy:", accuracy)
```

```python
# TO DO LIST TASK 1

work_list = []


def add_work():
    task_name = input("Type a task to add: ")
    work_list.append(task_name)
    print(f"Task '{task_name}' has been added to your list.")


def view_work():
    if len(work_list) == 0:
        print("Your task list is empty.")
    else:
        print("\nHere are your current tasks:")
        for number, name in enumerate(work_list, start=1):
            print(f"{number}. {name}")


def remove_work():
    view_work()
    if not work_list:
        return
    try:
        task_no = int(input("Enter the number of the task to remove: "))
        if 1 <= task_no <= len(work_list):
            removed_task = work_list.pop(task_no - 1)
            print(f"'{removed_task}' has been successfully removed.")
        else:
            print("No task found with that number.")
    except ValueError:
        print("Invalid input. Please enter a valid number.")


if __name__ == "__main__":
    print("Welcome to the Simple Task Manager")

    while True:
        print("\nSelect an option from the menu below:")
        print("-------------------------------------")
        print("1. Add a task")
        print("2. Remove a task")
        print("3. View all tasks")
        print("4. Close the program")

        user_input = input("Enter your choice: ")

        if user_input == "1":
            add_work()
        elif user_input == "2":
            remove_work()
        elif user_input == "3":
            view_work()
        elif user_input == "4":
            print("Program closed. Keep up the good work!")
            break
        else:
            print("Invalid option. Please try again.")
```

```python
import random

tries = 5

for round_num in range(1, tries + 1):
    print(f"\n--- Round {round_num} ---")

    comp_choice = random.randint(1, 20)
    print("Computer picked:", comp_choice)

    user_value = input("Enter your number: ")

    if user_value.isdigit():
        user_choice = int(user_value)

        total_sum = comp_choice + user_choice
        print(f"Total = {comp_choice} + {user_choice} = {total_sum}")

        if total_sum % 3 == 0 and total_sum % 5 == 0:
            print("FizzBuzz")
        elif total_sum % 3 == 0:
            print("Fizz")
        elif total_sum % 5 == 0:
            print("Buzz")
        else:
            print(total_sum)
    else:
        print("Invalid input.")

```

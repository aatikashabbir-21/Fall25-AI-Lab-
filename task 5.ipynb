```python
def depth_first_stack(graph_data, start_point):
    seen = set()
    stack_list = [start_point]

    while stack_list:
        item = stack_list.pop()
        if item not in seen:
            print(item, end=" ")
            seen.add(item)
            for link in graph_data[item]:
                if link not in seen:
                    stack_list.append(link)


graph_data = {
    '0': ['1', '2'],
    '1': ['3', '4'],
    '2': [],
    '3': [],
    '4': []
}

print("Depth First Search using Stack:")
depth_first_stack(graph_data, '0')
```

Tree Traversal in DFS (Depth First Search)
Tree traversal means visiting each part of a binary tree step by step.
There are three main DFS types:

Preorder (Root → Left → Right)

Visit the root first

Then go to the left side

Then the right side

Inorder (Left → Root → Right)

Visit left side first

Then the root

Then the right side
🔹 In a BST, this gives sorted values.

Postorder (Left → Right → Root)

Visit left side

Then right side

Then the root
🔹 Used when deleting a tree.

💻 Modified Code (No “node”, short variables)

#### implementation 

```python
tree = {
    '1': ['2', '3'],
    '2': ['4', '5'],
    '3': [None, None],
    '4': [None, None],
    '5': [None, None]
}

def go_inorder(tree_map, value):
    if value:
        go_inorder(tree_map, tree_map[value][0])
        print(value, end=" ")
        go_inorder(tree_map, tree_map[value][1])

def go_preorder(tree_map, value):
    if value:
        print(value, end=" ")
        go_preorder(tree_map, tree_map[value][0])
        go_preorder(tree_map, tree_map[value][1])

def go_postorder(tree_map, value):
    if value:
        go_postorder(tree_map, tree_map[value][0])
        go_postorder(tree_map, tree_map[value][1])
        print(value, end=" ")

print("Inorder:", end=" ")
go_inorder(tree, '1')

print("\nPreorder:", end=" ")
go_preorder(tree , '1')

print("\nPostorder:", end=" ")
go_postorder(tree , '1')
```
